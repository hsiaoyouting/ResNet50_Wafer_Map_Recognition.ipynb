### PyTorch Vision Transformer (ViT) 模型定義

這部分程式碼定義了 Vision Transformer 的核心架構。它將圖像分割成補丁 (patches)，為這些補丁添加位置信息，然後通過一系列 Transformer 編碼器層處理，最後通過一個分類頭進行預測。


````python
import torch
import torch.nn as nn

# --- 1. 定義 Patch Embedding 層 ---
class PatchEmbedding(nn.Module):
    def __init__(self, img_size, patch_size, in_channels, embed_dim):
        super().__init__()
        self.img_size = img_size
        self.patch_size = patch_size
        self.n_patches = (img_size // patch_size) ** 2 # 計算總共有多少個 patches

        # 2D 卷積層，用於將圖像分割成 patches 並進行線性投影
        # kernel_size 和 stride 都設為 patch_size，實現不重疊的 patch 提取
        self.proj = nn.Conv2d(
            in_channels, embed_dim, kernel_size=patch_size, stride=patch_size
        )

    def forward(self, x):
        # x 的形狀: (batch_size, in_channels, img_size, img_size)
        x = self.proj(x)  # 輸出形狀: (batch_size, embed_dim, n_patches_h, n_patches_w)
        # 將空間維度 (H, W) 展平，並將 embed_dim 移到最後一維
        x = x.flatten(2)  # 輸出形狀: (batch_size, embed_dim, n_patches)
        x = x.transpose(1, 2) # 輸出形狀: (batch_size, n_patches, embed_dim)
        return x

# --- 2. 定義多頭自注意力機制 (Multi-Head Self-Attention) ---
class MultiHeadSelfAttention(nn.Module):
    def __init__(self, embed_dim, num_heads=8, dropout=0.1):
        super().__init__()
        self.embed_dim = embed_dim
        self.num_heads = num_heads
        self.head_dim = embed_dim // num_heads # 每個注意力頭的維度
        self.scale = self.head_dim ** -0.5    # 縮放因子

        self.qkv = nn.Linear(embed_dim, embed_dim * 3) # 用於生成 Query, Key, Value
        self.proj = nn.Linear(embed_dim, embed_dim)     # 輸出投影層
        self.dropout = nn.Dropout(dropout)

    def forward(self, x):
        batch_size, n_patches, _ = x.shape
        # 生成 Q, K, V，並重塑以適應多頭注意力
        # (batch_size, n_patches, embed_dim * 3) -> (batch_size, n_patches, 3, num_heads, head_dim)
        qkv = self.qkv(x).reshape(batch_size, n_patches, 3, self.num_heads, self.head_dim)
        # 分離 Q, K, V，並調整維度使其變成 (batch_size, num_heads, n_patches, head_dim)
        q, k, v = qkv.permute(2, 0, 3, 1, 4)

        # 計算注意力分數 (Query @ Key.T / scale)
        attn_scores = (q @ k.transpose(-2, -1)) * self.scale
        attn = attn_scores.softmax(dim=-1)
        attn = self.dropout(attn)

        # 將注意力應用於 Value (attn @ V)
        # (batch_size, num_heads, n_patches, head_dim)
        weighted_avg = (attn @ v).transpose(1, 2).reshape(batch_size, n_patches, self.embed_dim)

        # 最終輸出投影
        out = self.proj(weighted_avg)
        return out

# --- 3. 定義前饋神經網路 (Feed-Forward Network) ---
class MLP(nn.Module):
    def __init__(self, embed_dim, mlp_dim, dropout=0.1):
        super().__init__()
        self.fc1 = nn.Linear(embed_dim, mlp_dim)
        self.gelu = nn.GELU()
        self.dropout = nn.Dropout(dropout)
        self.fc2 = nn.Linear(mlp_dim, embed_dim)

    def forward(self, x):
        x = self.fc1(x)
        x = self.gelu(x)
        x = self.dropout(x)
        x = self.fc2(x)
        x = self.dropout(x)
        return x

# --- 4. 定義 Transformer Encoder Block (一個編碼器層) ---
class TransformerEncoderBlock(nn.Module):
    def __init__(self, embed_dim, num_heads, mlp_dim, dropout=0.1):
        super().__init__()
        self.norm1 = nn.LayerNorm(embed_dim)
        self.attn = MultiHeadSelfAttention(embed_dim, num_heads, dropout)
        self.norm2 = nn.LayerNorm(embed_dim)
        self.mlp = MLP(embed_dim, mlp_dim, dropout)

    def forward(self, x):
        # LayerNorm -> Multi-Head Self-Attention -> Residual Connection
        x = x + self.attn(self.norm1(x))
        # LayerNorm -> MLP -> Residual Connection
        x = x + self.mlp(self.norm2(x))
        return x

# --- 5. 定義完整的 Vision Transformer (ViT) 模型 ---
class VisionTransformer(nn.Module):
    def __init__(self, img_size=224, patch_size=16, in_channels=3, num_classes=10,
                 embed_dim=768, depth=12, num_heads=12, mlp_dim=3072, dropout=0.1):
        super().__init__()

        self.patch_embed = PatchEmbedding(img_size, patch_size, in_channels, embed_dim)
        self.cls_token = nn.Parameter(torch.zeros(1, 1, embed_dim)) # 可學習的分類 token
        self.pos_embed = nn.Parameter(torch.zeros(1, 1 + self.patch_embed.n_patches, embed_dim)) # 位置嵌入
        self.dropout = nn.Dropout(dropout)

        # 堆疊多個 Transformer Encoder Block
        self.transformer_blocks = nn.ModuleList([
            TransformerEncoderBlock(embed_dim, num_heads, mlp_dim, dropout)
            for _ in range(depth)
        ])

        self.norm = nn.LayerNorm(embed_dim)
        self.head = nn.Linear(embed_dim, num_classes) # 分類頭

    def forward(self, x):
        batch_size = x.shape[0]

        # 1. Patch Embedding
        x = self.patch_embed(x) # (batch_size, n_patches, embed_dim)

        # 2. 添加 [CLS] token
        # 複製 [CLS] token 以匹配當前批次大小，並將其與 patch embeddings 拼接
        cls_tokens = self.cls_token.expand(batch_size, -1, -1)
        x = torch.cat((cls_tokens, x), dim=1) # (batch_size, 1 + n_patches, embed_dim)

        # 3. 添加位置嵌入
        x = x + self.pos_embed
        x = self.dropout(x)

        # 4. 經過 Transformer Encoder Blocks
        for block in self.transformer_blocks:
            x = block(x)

        # 5. 最終的 Layer Normalization 和分類頭
        # 我們只取 [CLS] token 的輸出進行分類 (位於序列的第一個位置)
        x = self.norm(x)
        cls_token_output = x[:, 0] # 取 [CLS] token 的輸出
        logits = self.head(cls_token_output)

        return logits

print("Vision Transformer 模型架構已定義完成。")

# --- 測試模型 (可選) ---
# device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
# # 假設有 10 個類別
# model = VisionTransformer(img_size=image_size, in_channels=3, num_classes=len(train_dataset.classes) if train_dataset else 2).to(device)
# print(f"模型總參數量：{sum(p.numel() for p in model.parameters() if p.requires_grad)}")

# # 創建一個隨機輸入張量
# dummy_input = torch.randn(1, 3, image_size, image_size).to(device)
# output = model(dummy_input)
# print(f"單個虛擬輸入的輸出 logits 形狀：{output.shape}")
# assert output.shape == (1, len(train_dataset.classes) if train_dataset else 2)
````

### PyTorch 模型訓練設定與訓練循環

這部分程式碼將配置訓練環境，包括選擇運算設備、初始化模型、定義損失函數和優化器，並提供一個基本的訓練和驗證循環。


````python
import torch
import torch.nn as nn
import torch.optim as optim
from tqdm.notebook import tqdm

# --- 1. 設定運算裝置 ---
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"使用裝置：{device}")

# --- 2. 實例化模型 ---
# 確保 train_dataset 在此處是可用的，否則 num_classes 預設為 2
# 如果之前的資料載入失敗，這裡可能會出錯。
num_classes = len(train_dataset.classes) if 'train_dataset' in locals() and train_dataset is not None else 2

# ViT 模型參數 (可以根據您的需求調整)
# 注意：這裡的 img_size 和 in_channels 應與資料預處理階段保持一致
model = VisionTransformer(
    img_size=image_size,      # 224 from data preprocessing
    patch_size=16,            # 每個 patch 的大小
    in_channels=3,            # 圖像輸入通道數 (RGB)
    num_classes=num_classes,  # 分類類別數量
    embed_dim=768,            # 嵌入維度
    depth=6,                 # Transformer 編碼器層數 (原版 ViT-B/16 是 12)
    num_heads=8,             # 多頭自注意力機制頭數 (原版 ViT-B/16 是 12)
    mlp_dim=3072,             # MLP 層的隱藏維度
    dropout=0.1
).to(device)

print(f"模型已實例化，並移至 {device}。")
print(f"模型總參數量：{sum(p.numel() for p in model.parameters() if p.requires_grad)}")

# --- 3. 定義損失函數和優化器 ---
criterion = nn.CrossEntropyLoss() # 交叉熵損失函數，適用於分類問題
optimizer = optim.AdamW(model.parameters(), lr=1e-4) # AdamW 優化器，學習率為 0.0001

print("損失函數和優化器已定義完成。")
````

````python
# --- 4. 定義訓練和驗證函數 ---
def train_epoch(model, dataloader, criterion, optimizer, device):
    model.train() # 設定模型為訓練模式
    running_loss = 0.0
    correct_predictions = 0
    total_samples = 0

    for images, labels in tqdm(dataloader, desc="Training"): # 使用 tqdm 顯示進度條
        images, labels = images.to(device), labels.to(device)

        optimizer.zero_grad() # 清除之前的梯度
        outputs = model(images) # 前向傳播
        loss = criterion(outputs, labels) # 計算損失
        loss.backward() # 後向傳播，計算梯度
        optimizer.step() # 更新模型參數

        running_loss += loss.item() * images.size(0)
        _, predicted = torch.max(outputs.data, 1)
        total_samples += labels.size(0)
        correct_predictions += (predicted == labels).sum().item()

    epoch_loss = running_loss / total_samples
    epoch_acc = correct_predictions / total_samples
    return epoch_loss, epoch_acc

def validate_epoch(model, dataloader, criterion, device):
    model.eval() # 設定模型為評估模式
    running_loss = 0.0
    correct_predictions = 0
    total_samples = 0

    with torch.no_grad(): # 在驗證階段不計算梯度
        for images, labels in tqdm(dataloader, desc="Validation"): # 使用 tqdm 顯示進度條
            images, labels = images.to(device), labels.to(device)

            outputs = model(images) # 前向傳播
            loss = criterion(outputs, labels) # 計算損失

            running_loss += loss.item() * images.size(0)
            _, predicted = torch.max(outputs.data, 1)
            total_samples += labels.size(0)
            correct_predictions += (predicted == labels).sum().item()

    epoch_loss = running_loss / total_samples
    epoch_acc = correct_predictions / total_samples
    return epoch_loss, epoch_acc

print("訓練和驗證函數已定義。")
````

````python
# --- 5. 執行訓練循環 ---
num_epochs = 5 # 訓練的 epoch 數量 (可以根據需求調整)

if 'train_loader' in locals() and train_loader is not None and \
   'val_loader' in locals() and val_loader is not None:
    print(f"開始訓練 {num_epochs} 個 Epochs...")
    for epoch in range(num_epochs):
        train_loss, train_acc = train_epoch(model, train_loader, criterion, optimizer, device)
        val_loss, val_acc = validate_epoch(model, val_loader, criterion, device)

        print(f"Epoch {epoch+1}/{num_epochs} | "
              f"Train Loss: {train_loss:.4f}, Train Acc: {train_acc:.4f} | "
              f"Val Loss: {val_loss:.4f}, Val Acc: {val_acc:.4f}")
    print("訓練完成！")
else:
    print("無法開始訓練，因為資料載入器未成功創建。請確保之前的資料載入步驟已成功執行。")
````

### 重新執行訓練循環

由於之前的訓練輸出可能被截斷，我們將重新執行訓練循環，以顯示所有 epoch 的訓練和驗證損失變化。這將從頭開始模型的訓練過程。

````python
# --- 5. 重新執行訓練循環 --- (為顯示完整輸出)
num_epochs = 5 # 訓練的 epoch 數量 (保持不變)

if 'train_loader' in locals() and train_loader is not None and \
   'val_loader' in locals() and val_loader is not None:
    print(f"重新開始訓練 {num_epochs} 個 Epochs 以顯示完整輸出...")
    # 需要重新初始化模型，否則會從上次訓練的狀態繼續，導致學習率下降等問題，或梯度累積。
    # 為了清晰地重新展示訓練過程，我們在此重新實例化模型。
    # 重新實例化模型和優化器
    model = VisionTransformer(
        img_size=image_size,      # 224 from data preprocessing
        patch_size=16,            # 每個 patch 的大小
        in_channels=3,            # 圖像輸入通道數 (RGB)
        num_classes=num_classes,  # 分類類別數量
        embed_dim=768,            # 嵌入維度
        depth=6,                 # Transformer 編碼器層數
        num_heads=8,             # 多頭自注意力機制頭數
        mlp_dim=3072,            # MLP 層的隱藏維度
        dropout=0.1
    ).to(device)
    optimizer = optim.AdamW(model.parameters(), lr=1e-4) # 重新初始化優化器

    # 初始化列表以儲存每個 epoch 的數據
    train_losses = []
    train_accuracies = []
    val_losses = []
    val_accuracies = []

    for epoch in range(num_epochs):
        train_loss, train_acc = train_epoch(model, train_loader, criterion, optimizer, device)
        val_loss, val_acc = validate_epoch(model, val_loader, criterion, device)

        # 儲存數據
        train_losses.append(train_loss)
        train_accuracies.append(train_acc)
        val_losses.append(val_loss)
        val_accuracies.append(val_acc)

        print(f"Epoch {epoch+1}/{num_epochs} | "
              f"Train Loss: {train_loss:.4f}, Train Acc: {train_acc:.4f} | "
              f"Val Loss: {val_loss:.4f}, Val Acc: {val_acc:.4f}")
    print("訓練完成！")
else:
    print("無法開始訓練，因為資料載入器未成功創建。請確保之前的資料載入步驟已成功執行。")
````

````python
# 定義資料路徑 (請替換為您實際的資料集路徑)
train_data_path = './data/wafer_defects/train'
val_data_path = './data/wafer_defects/val'

# ViT 圖像大小 (通常為 224x224 或類似解析度)
image_size = 224

# 為了方便測試，如果資料夾不存在，我們創建它們並生成一些虛擬圖像。
# 在實際應用中，您應該將此部分替換為您的真實資料集。
import os
from PIL import Image
import numpy as np

def create_dummy_images(path, num_classes=2, images_per_class=5, img_size=image_size):
    if not os.path.exists(path):
        os.makedirs(path)

    # Check if the directory is already populated
    if any(os.listdir(path)):
        print(f"資料夾 '{path}' 已有內容，跳過生成虛擬圖像。")
        return

    print(f"為 '{path}' 創建虛擬圖像...")
    for class_idx in range(num_classes):
        class_dir = os.path.join(path, f'class_{class_idx}')
        os.makedirs(class_dir, exist_ok=True)
        for i in range(images_per_class):
            # 創建一個隨機的 RGB 圖像
            dummy_image = Image.fromarray(np.random.randint(0, 255, (img_size, img_size, 3), dtype=np.uint8))
            dummy_image.save(os.path.join(class_dir, f'dummy_image_{i}.png'))
    print(f"完成為 '{path}' 創建虛擬圖像。")

# 執行創建虛擬圖像的函數
create_dummy_images(train_data_path, img_size=image_size)
create_dummy_images(val_data_path, img_size=image_size)

print("資料路徑已定義並檢查/創建虛擬資料集完成。")
````
