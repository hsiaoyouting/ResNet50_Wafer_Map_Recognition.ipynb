```python
import torch
import torch.nn as nn
import torch.optim as optim
import timm

# 1. 建立正確的 38 類 ViT 模型 (多類別分類)
class WaferViT38(nn.Module):
    def __init__(self, num_classes=38):
        super(WaferViT38, self).__init__()
        # 使用 timm 載入 ViT，設定 num_classes=38
        self.model = timm.create_model('vit_tiny_patch16_224', pretrained=False, num_classes=num_classes)

        # 修改第一層 Patch Embedding 以接受單通道 (1 channel) 晶圓圖
        old_conv = self.model.patch_embed.proj
        self.model.patch_embed.proj = nn.Conv2d(
            in_channels=1,
            out_channels=old_conv.out_channels,
            kernel_size=old_conv.kernel_size,
            stride=old_conv.stride,
            padding=old_conv.padding
        )

    def forward(self, x):
        # 調整尺寸以符合 ViT 預設的 224x224 輸入
        if x.shape[2] != 224 or x.shape[3] != 224:
            x = nn.functional.interpolate(x, size=(224, 224), mode='bilinear', align_corners=False)
        return self.model(x)

model = WaferViT38(num_classes=38).to(device)

# 2. 38 類多類別分類必須使用 CrossEntropyLoss
criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.parameters(), lr=1e-4)

# 3. 訓練迴圈
epochs = 5
for epoch in range(epochs):
    model.train()
    running_loss = 0.0

    for inputs, labels in train_loader:
        inputs, labels = inputs.to(device), labels.to(device)

        # 確保 labels 是一維的類別索引 (LongTensor)，若原本是 One-hot 需先 argmax
        if len(labels.shape) > 1 and labels.shape[1] > 1:
            labels = torch.argmax(labels, dim=1)
        labels = labels.long()

        optimizer.zero_grad()
        outputs = model(inputs)
        loss = criterion(outputs, labels)
        loss.backward()
        optimizer.step()

        running_loss += loss.item() * inputs.size(0)

    epoch_loss = running_loss / len(train_loader.dataset)
    print(f"Epoch [{epoch+1}/{epochs}], Loss: {epoch_loss:.4f}")

print("模型訓練完成！請再執行評估與畫圖程式碼。")
```

```python
import numpy as np
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
import pandas as pd
import pickle
import timm

import warnings
warnings.filterwarnings('ignore')

# 檢查是否有 GPU 可用
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f"Using device: {device}")
```

```python
class WaferDataset(Dataset):
    def __init__(self, x_data, y_data):
        # PyTorch 的影像通道維度需為 (Channels, Height, Width)
        # 原本資料 shape 為 (N, H, W) 或 (N, H, W, 1)，需調整為 (N, 1, H, W)
        if len(x_data.shape) == 3:
            x_data = np.expand_dims(x_data, axis=1)
        elif x_data.shape[-1] == 1:
            x_data = np.transpose(x_data, (0, 3, 1, 2))

        self.x_data = torch.tensor(x_data, dtype=torch.float32)
        self.y_data = torch.tensor(y_data, dtype=torch.float32)

    def __len__(self):
        return len(self.x_data)

    def __getitem__(self, idx):
        return self.x_data[idx], self.y_data[idx]


class data_loader:
    def __init__(self, dataset_path, vis=False):
        """初始化 data_loader 類別"""
        self.dataset_path = dataset_path
        self.is_vis = vis

        # 標籤對應的缺陷類型
        self.label_keys = ["Center, ", "Donut, ", "Edge_Loc, ", "Edge_Ring, ",
                           "Loc, ", "Near_Full, ", "Scratch, ", "Random, "]

    def load_data(self):
        """載入資料"""
        self.data = np.load(self.dataset_path)
        self.train = self.data["arr_0"]
        self.label = self.data["arr_1"]
        print("MixedWM38:", np.shape(self.label)[0], 'wafer loaded')

    def prep_data(self, batch_size=32):
        """進行 train_test_split 並建立 PyTorch DataLoader"""
        self.x_train, self.x_test, self.y_train, self.y_test = train_test_split(
            self.train, self.label, test_size=0.2
        )

        # 建立 Dataset 與 DataLoader
        train_dataset = WaferDataset(self.x_train, self.y_train)
        test_dataset = WaferDataset(self.x_test, self.y_test)

        self.train_loader = DataLoader(train_dataset, batch_size=batch_size, shuffle=True)
        self.test_loader = DataLoader(test_dataset, batch_size=batch_size, shuffle=False)

    def read_label(self, label, defect_types=''):
        """將標籤轉換為缺陷字串"""
        if np.sum(label) == 0:
            defect_types = 'Normal wafer'
        else:
            for digit in range(np.shape(label)[0]):
                if label[digit] == 1:
                    defect_types = defect_types + self.label_keys[digit]
        return defect_types

    def see_wafer(self, wafer_num):
        """檢視指定編號的晶圓缺陷圖"""
        defect_types = self.read_label(self.label[wafer_num])

        plt.title("wafer #" + str(wafer_num))
        plt.imshow(self.train[wafer_num])
        plt.colorbar()
        plt.show()

        print("Defect types =", defect_types)
        print("Labeled as:", self.label[wafer_num])

    def get_data(self, batch_size=32):
        """取得資料載入器"""
        self.load_data()
        self.prep_data(batch_size=batch_size)

        return self.train_loader, self.test_loader
```

```python
dataset_path = 'Wafer_Map_Datasets.npz'  # 請依實際路徑調整

A1 = data_loader(dataset_path, vis=True)
train_loader, test_loader = A1.get_data(batch_size=32)
```

```python
class WaferViT(nn.Module):
    def __init__(self, num_classes=8):
        super(WaferViT, self).__init__()
        # 使用 timm 載入輕量級的 vit_tiny_patch16_224 或其他 ViT 模型
        # pretrained=False (若是灰階小圖，通常從頭訓練比較合適)
        self.model = timm.create_model('vit_tiny_patch16_224', pretrained=False, num_classes=num_classes)

        # 修改第一層 Patch Embedding 以接受單通道 (1 channel) 輸入
        # 原本 ViT 通常預設輸入為 3 通道 (RGB)
        old_conv = self.model.patch_embed.proj
        self.model.patch_embed.proj = nn.Conv2d(
            in_channels=1,
            out_channels=old_conv.out_channels,
            kernel_size=old_conv.kernel_size,
            stride=old_conv.stride,
            padding=old_conv.padding
        )

    def forward(self, x):
        # 由於 ViT 預設輸入尺寸較大 (如 224x224)，如果晶圓圖尺寸較小 (例如 26x26)，
        # timm 的 patch embedding 會自動處理，但建議確認尺寸或透過 interpolate 放大
        if x.shape[2] != 224 or x.shape[3] != 224:
            x = nn.functional.interpolate(x, size=(224, 224), mode='bilinear', align_corners=False)
        return self.model(x)

model = WaferViT(num_classes=8).to(device)
print(model)
```

```python
# 由於是多標籤分類 (Multi-label)，使用 BCEWithLogitsLoss
criterion = nn.BCEWithLogitsLoss()
optimizer = optim.Adam(model.parameters(), lr=1e-4)

epochs = 5

for epoch in range(epochs):
    model.train()
    running_loss = 0.0

    for inputs, labels in train_loader:
        inputs, labels = inputs.to(device), labels.to(device)

        optimizer.zero_grad()
        outputs = model(inputs)
        loss = criterion(outputs, labels)
        loss.backward()
        optimizer.step()

        running_loss += loss.item() * inputs.size(0)

    epoch_loss = running_loss / len(train_loader.dataset)
    print(f"Epoch [{epoch+1}/{epochs}], Loss: {epoch_loss:.4f}")

print("訓練完成！")
```

```python
from sklearn.metrics import multilabel_confusion_matrix, classification_report
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
import torch

model.eval()
all_preds = []
all_targets = []

with torch.no_grad():
    for inputs, labels in test_loader:
        inputs = inputs.to(device)
        outputs = model(inputs)

        preds = (torch.sigmoid(outputs) > 0.5).int().cpu().numpy()
        true_labels = labels.int().cpu().numpy()

        all_preds.extend(preds)
        all_targets.extend(true_labels)

all_preds = np.array(all_preds)
all_targets = np.array(all_targets)

# 印出分類報告
print("Multi-label Classification Report:")
print(classification_report(all_targets, all_preds, zero_division=0))

# 針對每個類別分別計算並繪製混淆矩陣
mcm = multilabel_confusion_matrix(all_targets, all_preds)
num_classes = mcm.shape[0]

fig, axes = plt.subplots(2, (num_classes + 1) // 2, figsize=(15, 8))
axes = axes.flatten()

for i in range(num_classes):
    sns.heatmap(mcm[i], annot=True, fmt='d', cmap='Blues', ax=axes[i], cbar=False)
    axes[i].set_title(f'Class {i} Confusion Matrix')
    axes[i].set_xlabel('Predicted')
    axes[i].set_ylabel('True')

plt.tight_layout()
plt.show()
```
